# Dataset 1 — Appliances Energy Prediction (UCI)
 Situação

Uma empresa de eficiência energética está analisando o comportamento de uma residência de baixo consumo. A equipe deseja identificar períodos de consumo elevado dos eletrodomésticos e observar quais condições de temperatura e umidade estavam presentes nesses momentos.  

```python
# 1.Carregue a amostra no Pandas e apresente head(), shape, info() e describe().
dataset_1 = pd.read_csv('energy_data_complete.csv')

dataset_1.head()
```

```python
dataset_1.shape
```

```python
dataset_1.info()
```

```python
dataset_1.describe()
```

```python
# 2.Renomeie Appliances para Consumo_Eletrodomesticos e simplifique o nome de pelo menos três atributos ambientais.
dataset_1.rename(columns={'Appliances': 'Consumo_Eletrodomesticos',
                          'T1': 'Temperatura_1',
                          'T2': 'Temperatura_2',
                          'T3': 'Temperatura_3'}, inplace = True)
```

```python
# 3.Determine o maior consumo de eletrodomésticos registrado na amostra.
maior_consumo_1 = dataset_1['Consumo_Eletrodomesticos'].max()

print(f'O maior consumo de eletrodomésticos registrado na amostra foi de: {maior_consumo_1} Wh')
```

```python
# 4.Calcule um limiar correspondente a 70% do valor máximo e crie um DataFrame contendo somente os registros acima desse limite.
limiar_70_1 = maior_consumo_1 * 0.70

print(f'O limiar de 70% do consumo máximo é: {limiar_70_1} Wh')

dataset_acima_70_1 = dataset_1[
    dataset_1['Consumo_Eletrodomesticos'] > limiar_70_1]

print(dataset_acima_70_1.head())
```

```python
# 5.Conte os registros selecionados e calcule qual percentual da amostra eles representam.
quantidade_acima_70_1 = dataset_acima_70_1.shape[0]

quantidade_total_1 = dataset_1.shape[0]

percentual_acima_70_1 = (quantidade_acima_70_1 / quantidade_total_1) * 100

print(f'Quantidade de registros acima de 70%: {quantidade_acima_70_1}')
print(f'Quantidade total de registros: {quantidade_total_1}')
print(f'Percentual da amostra: {percentual_acima_70_1:.2f}%')
```

```python
# 6.Calcule a temperatura média de T1 e crie outro DataFrame
# contendo simultaneamente consumo acima de 70% do máximo e temperatura acima da média.
temperatura_media_1 = dataset_1['Temperatura_1'].mean()

print(f'A temperatura média de T1 foi: {temperatura_media_1:.2f} °C')
```

```python
dataset_acima_70_e_temp_media_1 = dataset_1[
    (dataset_1['Consumo_Eletrodomesticos'] > limiar_70_1) &
    (dataset_1['Temperatura_1'] > temperatura_media_1)]

print(dataset_acima_70_e_temp_media_1.head())
```

```python
# 7.Compare os dois DataFrames e explique o efeito da inclusão da temperatura como segundo critério.
registro_1_1 = dataset_acima_70_1.shape[0]
registro_2_1 = dataset_acima_70_e_temp_media_1.shape[0]

print(f'Registros com consumo acima de 70%: {registro_1_1}')
print(f'Registros com consumo acima de 70% e temperatura acima da média: {registro_2_1}')

print(f'Registros eliminados pela inclusão da temperatura: {registro_1_1 - registro_2_1}')
```

# Dataset 3 — Power Consumption of Tetouan City (UCI)

Situação

Uma equipe responsável pelo planejamento da distribuição de energia em Tétouan deseja identificar qual das três zonas apresenta o maior pico de consumo e observar as condições ambientais registradas nos momentos de maior demanda.

```python
dataset_3 = pd.read_csv('Tetuan City power consumption.csv')
```

```python
# 1.Renomeie as três variáveis de consumo
# para Consumo_Zona_1, Consumo_Zona_2 e Consumo_Zona_3
dataset_3.rename(columns={
                          'Zone 1 Power Consumption': 'Consumo_Zona_1',
                          'Zone 3  Power Consumption': 'Consumo_Zona_3',
                          'Zone 2  Power Consumption': 'Consumo_Zona_2',
                          'Temperature': 'Temperatura',
                          'Humidity': 'Umidade',
                          'Wind Speed': 'Velocidade_Vento',}, inplace=True)
```

```python
# 2.Determine o consumo máximo registrado em cada uma das três zonas.
max_zona_1_3 = dataset_3['Consumo_Zona_1'].max()
max_zona_2_3 = dataset_3['Consumo_Zona_2'].max()
max_zona_3_3 = dataset_3['Consumo_Zona_3'].max()

print(f'Máximo da Zona 1: {max_zona_1_3:.2f}')
print(f'Máximo da Zona 2: {max_zona_2_3:.2f}')
print(f'Máximo da Zona 3: {max_zona_3_3:.2f}')
```

```python
# 3.Identifique qual zona apresenta o maior pico de consumo na amostra.
maximos_3 = {
    'Consumo_Zona_1': max_zona_1_3,
    'Consumo_Zona_2': max_zona_2_3,
    'Consumo_Zona_3': max_zona_3_3
}

zona_maior_pico_3 = max(maximos_3, key=maximos_3.get)
maior_pico_3 = maximos_3[zona_maior_pico_3]

print(f'A zona com maior pico de consumo é: {zona_maior_pico_3}')
print(f'O maior pico registrado foi: {maior_pico_3:.2f}')
```

```python
# 4.Para a zona identificada, calcule 70% do valor máximo e
# crie um DataFrame com os registros acima desse limiar.
limiar_70_3 = maior_pico_3 * 0.70

print(f'O limiar de 70% do maior pico é: {limiar_70_3:.2f}')
```

```python
dataset_consumo_elevado_3 = dataset_3[
    dataset_3[zona_maior_pico_3] > limiar_70_3]

dataset_consumo_elevado_3.head()
```

```python
# 5.Conte os registros e calcule o percentual que eles representam.
quantidade_elevada_3 = dataset_consumo_elevado_3.shape[0]

quantidade_total_3 = dataset_3.shape[0]

percentual_elevado_3 = (quantidade_elevada_3 / quantidade_total_3) * 100

print(f'Quantidade de registros acima de 70%: {quantidade_elevada_3}')
print(f'Quantidade total de registros: {quantidade_total_3}')
print(f'Percentual da amostra: {percentual_elevado_3:.2f}%')
```

```python
# 6.Calcule a temperatura média e crie um segundo DataFrame
# contendo consumo acima do limiar e temperatura acima da média.
temperatura_media_3 = dataset_3['Temperatura'].mean()

print(f'A temperatura média da amostra é: {temperatura_media_3:.2f} °C')
```

```python
dataset_consumo_e_temperatura_3 = dataset_3[
    (dataset_3[zona_maior_pico_3] > limiar_70_3) &
    (dataset_3['Temperatura'] > temperatura_media_3)
]
quantidade_segundo_3 = dataset_consumo_e_temperatura_3.shape[0]

percentual_segundo_3 = (quantidade_segundo_3 / quantidade_total_3) * 100

print(f'Quantidade de registros: {quantidade_segundo_3}')
print(f'Percentual da amostra: {percentual_segundo_3:.2f}%')

dataset_consumo_e_temperatura_3.head()
```

```python
# 7.Compare os dois conjuntos e explique o que ocorreu com a quantidade de registros
# após a inclusão da condição ambiental.
print(f'Registros com consumo acima de 70%: {quantidade_elevada_3}')

print(f'Registros com consumo acima de 70% e temperatura acima da média: {quantidade_segundo_3}')

print(f'Registros eliminados pela inclusão da temperatura: {quantidade_elevada_3 - quantidade_segundo_3}')
```

# Dataset 4 — Solar Power Generation Data (Kaggle)

Situação

Uma equipe de operação de uma usina fotovoltaica deseja localizar períodos de alta geração e identificar quais inversores aparecem com maior frequência nesses momentos. Nesta atividade, o foco inicial será o arquivo de geração da planta, sem realizar junção com o arquivo de sensores.

```python
# 1.Carregue a amostra e renomeie as principais variáveis para nomes
# mais claros, como Potencia_CC, Potencia_CA e Geracao_Diaria.
dataset_4 = pd.read_csv('Plant_1_Generation_Data.csv')

dataset_4.rename(columns={
    'DATE_TIME': 'Data_Hora',
    'SOURCE_KEY': 'Inversor',
    'DC_POWER': 'Potência_CC',
    'AC_POWER': 'Potência_CA',
    'DAILY_YIELD': 'Geração_Diaria',
    'TOTAL_YIELD': 'Geração_Total'}, inplace=True)
```

```python
# 2.Apresente head(), shape, info() e describe()

dataset_4.head()
```

```python
dataset_4.shape
```

```python
dataset_4.info()
```

```python
dataset_4.describe()
```

```python
# 3.Determine a maior potência CA registrada.
maior_potencia_ca_4 = dataset_4['Potência_CA'].max()

print(f'A maior potência CA registrada foi: {maior_potencia_ca_4:.2f}')
```

```python
# 4.Calcule um limiar correspondente a 70% da potência CA máxima.
limiar_70_4 = maior_potencia_ca_4 * 0.70

print(f'O limiar de 70% da potência CA máxima é: {limiar_70_4:.2f}')
```

```python
# 5.Crie um DataFrame contendo
# somente registros acima desse limite e calcule sua quantidade e percentual
dataset_alta_geracao_4 = dataset_4[
    dataset_4['Potência_CA'] > limiar_70_4]

dataset_alta_geracao_4.head()
```

```python
quantidade_alta_geracao_4 = dataset_alta_geracao_4.shape[0]

quantidade_total_4 = dataset_4.shape[0]

percentual_alta_geracao_4 = (quantidade_alta_geracao_4 / quantidade_total_4) * 100

print(f'Quantidade de registros de alta geração: {quantidade_alta_geracao_4}')

print(f'Quantidade total de registros: {quantidade_total_4}')

print(f'Percentual da amostra: {percentual_alta_geracao_4:.2f}%')
```

```python
# 6.Utilize value_counts() em SOURCE_KEY dentro do DataFrame de alta geração
dataset_alta_geracao_4['Inversor'].value_counts()
```

```python
#7.Identifique qual inversor aparece com maior frequência nos registros de alta geração
# e descreva o que esse resultado permite observar, sem concluir desempenho ou falha
# apenas com esse recorte.
inversor_mais_frequente_4 = (
    dataset_alta_geracao_4['Inversor']
    .value_counts()
    .idxmax()
)

quantidade_inversor_4 = (
    dataset_alta_geracao_4['Inversor']
    .value_counts()
    .max()
)

print(f'Inversor mais frequente nos registros de alta geração: {inversor_mais_frequente_4}')

print(f'Quantidade de ocorrências: {quantidade_inversor_4}')

print("O inversor que aparece com maior frequência no "
"conjunto de registros de alta geração é aquele que possui "
"maior participação nesse recorte específico da amostra."
"Esse resultado permite identificar quais inversores aparecem"
" mais frequentemente associados a períodos de potência CA "
"elevada.")
```

# Dataset 5 — Wind & Solar Energy Production (Kaggle)

Situação

Um operador de um portfólio de fontes renováveis deseja comparar a ocorrência de períodos de alta produção solar e alta produção eólica. Como as duas fontes possuem escalas próprias, cada uma deverá ser comparada com o seu próprio valor máximo.

```python
dataset_5 = pd.read_csv('Energy Production Dataset.csv')
```

```python
# 1.Renomeie as variáveis principais para Geracao_Solar e Geracao_Eolica.
dataset_solar_5 = dataset_5[
    dataset_5['Source'] == 'Solar']

dataset_eolica_5 = dataset_5[
    dataset_5['Source'] == 'Wind']
dataset_solar_5.rename(
    columns={'Production': 'Geracao_Solar'},
    inplace=True)

dataset_eolica_5.rename(
    columns={'Production': 'Geracao_Eolica'},
    inplace=True)
```

```python
# conferir
print('Solar:')
dataset_solar_5.head()
```

```python
print('Eólica:')
dataset_eolica_5.head()
```

```python
#2.Determine o valor máximo registrado para cada fonte
max_solar_5 = dataset_solar_5['Geracao_Solar'].max()

print(f'Máxima geração solar: {max_solar_5}')

max_eolica_5 = dataset_eolica_5['Geracao_Eolica'].max()

print(f'Máxima geração eólica: {max_eolica_5}')
```

```python
#3.Calcule separadamente 70% do máximo solar e 70% do máximo eólico.
limiar_solar_5 = max_solar_5 * 0.70
limiar_eolica_5 = max_eolica_5 * 0.70

print(f'Limiar de 70% da geração solar: {limiar_solar_5:.2f}')
print(f'Limiar de 70% da geração eólica: {limiar_eolica_5:.2f}')
```

```python
#4.Crie um DataFrame para registros de alta geração solar e outro para registros de alta geração eólica.
alta_solar_5 = dataset_solar_5[
    dataset_solar_5['Geracao_Solar'] > limiar_solar_5]

alta_solar_5.head()
```

```python
alta_eolica_5 = dataset_eolica_5[
    dataset_eolica_5['Geracao_Eolica'] > limiar_eolica_5]

alta_eolica_5.head()
```

```python
#5.Conte os registros de cada DataFrame e calcule seus respectivos percentuais.
quantidade_solar_5 = alta_solar_5.shape[0]

print(f'Registros de alta geração solar: {quantidade_solar_5}')

total_solar_5 = dataset_solar_5.shape[0]

percentual_solar_5 = (quantidade_solar_5 / total_solar_5) * 100

print(f'Percentual de alta geração solar: {percentual_solar_5:.2f}%')
```

```python
quantidade_eolica_5 = alta_eolica_5.shape[0]

print(f'Registros de alta geração eólica: {quantidade_eolica_5}')

total_eolica_5 = dataset_eolica_5.shape[0]

percentual_eolica_5 = (quantidade_eolica_5 / total_eolica_5) * 100

print(f'Percentual de alta geração eólica: {percentual_eolica_5:.2f}%')
```

```python
#6.Compare qual fonte aparece com maior frequência acima de 70% do seu próprio máximo.
if percentual_solar_5 > percentual_eolica_5:
    print('A geração solar aparece com maior frequência acima de 70% do próprio máximo.')

elif percentual_eolica_5 > percentual_solar_5:
    print('A geração eólica aparece com maior frequência acima de 70% do próprio máximo.')

else:
    print('As duas fontes apresentam a mesma frequência.')
```

```python
#7.Explique por que não seria adequado utilizar diretamente o mesmo valor numérico de potência como limite para as duas fontes sem antes observar a escala de cada variável.
print("Não seria adequado utilizar diretamente o mesmo valor numérico de potência como limite para as duas fontes porque a geração solar e a geração eólica podem apresentar escalas e valores máximos diferentes.")
```
